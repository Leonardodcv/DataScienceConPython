```python
import plotly.graph_objects as go
```

```python
fig10 = go.Figure()

fig10.add_trace(
    go.Bar(
        x=[1,2,3,4,5],
        y=[17, 10, 14, 15,18],
        name="barras"
    )
)

fig10.add_trace(
    go.Scatter(
        x=[1,2,3,4,5],
        y=[16,12,11,18,19],
        mode="lines",
        name="linea"
    )
)

layout = go.Layout(
    title="Grafico personalizado",
    xaxis_title="Eje X",
    yaxis_title="Eje Y"
)

fig10.update_layout(layout)

```

```python
# configuracion del grafico de linea
fig10.update_traces(
    selector = dict(name="linea"),
    line=dict(color="firebrick", width=4)
)
# configuracion del grafico de barras
fig10.update_traces(
    selector = dict(name="barras"),
    marker = dict(color="lightgreen")
)

fig10.update_layout(bargap=0.4)
```

```python
# Titulo de los graficos
fig10.update_layout(
    title = dict(
        text = "Este es mi grafico <br><sup>Subtitulo del grafico</sup>",
        font = dict(
            size=24,
            color="purple", 
            family="Times New Roman"
        ),
        x=0.5,
        xanchor="center"
    )
)

#Titulos de los Ejes
fig10.update_layout(
    xaxis_title=dict(font=dict(size=18, color="blue", family="Courier New")),
    yaxis_title=dict(font=dict(size=18, color="green", family="Courier New")),
)
```

```python
fig10.update_layout(
    xaxis=dict(
        showgrid=True,
        gridcolor="lightgray",
        zeroline=True,
        zerolinecolor="gray",
        showline=True,
        linecolor="black",
        linewidth=2
    ),
    yaxis = dict(
        gridcolor="lightgray",
        zeroline=True,
        zerolinecolor="gray",
        showline=True,
        linecolor="black",
        linewidth=2
    )
)
```

```python
fig10.update_layout(
    legend=dict(
        x=0.3,
        y=0.9,
        bgcolor="rgba(255,255,255,0.5)",
        bordercolor="black",
        borderwidth=2,
        font = dict(size=14, color="black", family="Arial")
    )
)
```

```python
#sirve para modificar el fondo de los graficos
fig10.update_layout(
    plot_bgcolor="rgba(123,65,789,0.5)",
    paper_bgcolor="lightblue"
)
```

```python
# modificar las anotaciones
fig10.add_annotation(
    x=2, y=12,
    text="Punto clave",
    showarrow=True,
    arrowhead=5,
    ax=20,
    ay=-30
)
```

```python
import dash
from dash import dcc, html
import pandas as pd
import plotly.express as px

# Crear un DataFrame de ejemplo
df = pd.DataFrame({
    "Fruta": ["Manzanas", "Naranjas", "Bananas", "Manzanas", "Naranjas", "Bananas"],
    "Cantidad": [4, 1, 2, 2, 4, 5],
    "Ciudad": ["BG", "BG", "BG", "BCN", "BCN", "BCN"]})

# Crear la aplicación Dash
app = dash.Dash(__name__)

# Definir el layout de la aplicación
app.layout = html.Div(children=[
    html.H1(children='Dashboard Interactivo Avanzado'),
    dcc.Dropdown(
        id='desplegable',
        options=[{'label': 'Bogotá', 'value': 'BG'},
                 {'label': 'Barcelona', 'value': 'BCN'}],
        value='BG'),
    dcc.Graph(id='grafico'),
])

# Callback para actualizar el gráfico basado en la selección del dropdown
@app.callback(
    dash.Output('grafico', 'figure'),
    [dash.Input('desplegable', 'value')]
)
def actualizar_grafico(ciudad_seleccionada):
    df_filtrado = df[df['Ciudad'] == ciudad_seleccionada]
    fig = px.bar(df_filtrado, x='Fruta', y='Cantidad', color='Fruta')
    return fig

# Ejecutar la aplicación
if __name__ == '__main__':
    app.run(debug=True)
```
